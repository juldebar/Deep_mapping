```python
for session in LIST_SESSION:

    session_path = Path(JULIEN_FOLDER, session)
    print(session_path)
    if not session_path.exists():
        print("ERROR SESSION NOT FOUND")
        continue

    metadata_df = pd.read_csv(Path(session_path, "METADATA", "metadata.csv"))

    for i, row in tqdm(metadata_df.iterrows(), total=len(metadata_df)):
        path_img = Path(session_path.parent, *[x for x in row["photo_relative_file_path"].replace("\\", "/").split("/") if x])

        try:
            with Image.open(path_img) as img:
                img.thumbnail((224,224))
        except OSError:
            print(path_img)
        
    
```
