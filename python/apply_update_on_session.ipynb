```python
from pathlib import Path
import shutil
import pandas as pd

JULIEN_FOLDER = Path("/scale/project/rummo/julien_maurice")
```

```python
# 20190524_MUS-Rodrigues-Saint-Francois_SCUBA-01_01
session_path = Path(JULIEN_FOLDER, "20190524_MUS-Rodrigues-Saint-Francois_SCUBA-01_01")
list_image_to_update = ["G0039895", "G0039880", "G0039881", "G0039882", "G0039883", "G0039884", "G0039885", "G0039886", "G0039887", "G0039888", "G0039889", "G0039890", "G0039891", "G0039892", "G0039893", "G0039894"] 

if not session_path.exists(): raise FileNotFound("Session Not found")
metadata_csv = Path(session_path, "METADATA", "metadata.csv")
metadata_df = pd.read_csv(metadata_csv)

def updateLatitude(row):
    if row["OriginalFileName"].replace(".JPG", "") in list_image_to_update and "181GOPRO" in row["photo_relative_file_path"]: 

        return row["GPSPosition"].split(" ")[0]
    return row["GPSLatitude"]


def updateLongitude(row):
    if row["OriginalFileName"].replace(".JPG", "") in list_image_to_update and "181GOPRO" in row["photo_relative_file_path"]: 
        return row["GPSPosition"].split(" ")[1]

        
    return row["GPSLongitude"]

    

metadata_df["GPSLatitude"] = metadata_df.apply(lambda row: updateLatitude(row), axis=1)
metadata_df["GPSLongitude"] = metadata_df.apply(lambda row: updateLongitude(row), axis=1)


metadata_df.to_csv(metadata_csv, index=False)
```
