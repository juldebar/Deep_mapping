```python
import shutil
import pandas as pd
from pathlib import Path
import traceback

from src.clean_and_rename import get_old_new_name

ROOT_FOLDER = Path("/scale/project/rummo/julien_maurice")

conversion_old_new_df = pd.read_csv("./csv/conversion_old_new_session_name.csv")
map_old_by_new = {row["new_session_name"]:row["old_session_name"] for i, row in conversion_old_new_df.iterrows()}
map_new_by_old = {row["old_session_name"]:row["new_session_name"] for i, row in conversion_old_new_df.iterrows()}
```

```python
cpt = 0
for session in sorted(list(ROOT_FOLDER.iterdir())):

    try:
        old_session_name, new_session_name = get_old_new_name(session.name, map_old_by_new, map_new_by_old)
        if old_session_name == None and new_session_name == None:
            print(f"Session {session.name} not found in mapping")
            continue

        print(f"\n\nWorking with {session.name}")
        if session.name == old_session_name:
            print("[WARNING] it's old session name")
            continue


        # Check PROCESSED_DATA_IA
        processed_data_folder = Path(session, "PROCESSED_DATA")
        ia_folder = Path(processed_data_folder, "IA")

        if processed_data_folder.exists() and len(list(processed_data_folder.iterdir())) == 1 and ia_folder.exists():
            print("Processed_data")
            #shutil.rmtree(processed_data_folder)
        
        # CHECK pdf preview
        preview_file = Path(session, f"000_{session.name}_preview.pdf")
        if preview_file.exists():
            print("preview file")
            #preview_file.unlink()

        leinfo = Path(session, "leinfo.sav")
        if leinfo.exists():
            print("leinfo exists")
            #leinfo.unlink()

        dcmi_file = Path(session, f"{session.name}_DCMI_metadata.csv")

        

        for file in session.iterdir():
            if "session" in file.name and file.is_file() and file.suffix == ".csv" and dcmi_file.exists():
                print("OLD DCMI.csv found", file, dcmi_file)
                # file.unlink()
                #shutil.move(file, dcmi_file)
                cpt += 1
            if "not" in file.name and file.is_dir():
                print("folder not wanted found: ", file)
                #shutil.rmtree(file)

            if "DCMI" not in file.name and file.is_file() and file.suffix == ".csv": 
                print(file)
    
    except Exception:
        # Print error
        print(traceback.format_exc(), end="\n\n")

print(cpt)
```
